## Serve this primitive as an endpoint

This archive ships its own `powerai_endpoint.py`, so the cells below run *that* —
not a Hub-written reference implementation — and open a public HTTPS tunnel to it.

Two things to know before you rely on it:

* **It dies with this session.** Colab reclaims idle notebooks after ~90 minutes and
  caps sessions around 12 hours. The Hub stores one address per primitive, so when this
  session ends that address answers nothing until you re-run and paste a new one.
* **Each person who runs this gets a different URL.** One Hub, one address — so this is
  a demo you drive, not something visitors wire up themselves.

The next cell downloads `cloudflared` (Cloudflare's official release), because a Colab
VM has no inbound networking of its own.


```python
import os, re, subprocess, sys, time, urllib.request

PORT = 8011
MODULE = "powerai_endpoint"
CLOUDFLARED = "/usr/local/bin/cloudflared"

# Re-running this cell has to be safe: a previous run leaves a server and a tunnel
# alive, and the next one would otherwise pass the health probe anyway because the
# *stale* server answers it — a stale process serving stale code, reported as fresh.
for pattern in ("cloudflared tunnel", f"uvicorn {MODULE}:app"):
    subprocess.run(["pkill", "-f", pattern], capture_output=True)
time.sleep(1)

if not os.path.exists(CLOUDFLARED):
    got = subprocess.run(
        "wget -q https://github.com/cloudflare/cloudflared/releases/latest/download/"
        f"cloudflared-linux-amd64 -O {CLOUDFLARED} && chmod +x {CLOUDFLARED}",
        shell=True, capture_output=True, text=True,
    )
    if got.returncode != 0:
        raise RuntimeError(f"could not install cloudflared:\n{got.stderr[-1000:]}")
    print("cloudflared installed")
else:
    print("cloudflared already present")

# --app-dir puts WORKDIR on the import path for this one process, so the archive
# need not be pip-installed (and may not even be a package) to be served.
server = subprocess.Popen(
    [sys.executable, "-m", "uvicorn", f"{MODULE}:app", "--app-dir", WORKDIR,
     "--host", "127.0.0.1", "--port", str(PORT)],
    stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL,
)

for _ in range(90):
    try:
        urllib.request.urlopen(f"http://127.0.0.1:{PORT}/health", timeout=1)
        break
    except Exception:
        time.sleep(1)
else:
    raise RuntimeError(
        f"{MODULE}.py did not come up as a healthy server on :{PORT} — check its "
        "own errors above; this is the archive's own code, not the Hub's"
    )
print(f"endpoint up on :{PORT}")

tunnel = subprocess.Popen(
    ["cloudflared", "tunnel", "--no-autoupdate", "--url", f"http://127.0.0.1:{PORT}"],
    stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1,
)
PUBLIC_URL = ""
deadline = time.time() + 90
while time.time() < deadline:
    line = tunnel.stdout.readline()
    if not line:
        break
    found = re.search(r"https://[-a-z0-9]+\.trycloudflare\.com", line)
    if found:
        PUBLIC_URL = found.group(0)
        break
if not PUBLIC_URL:
    raise RuntimeError("no tunnel URL; re-run this cell")

ENDPOINT_URL = PUBLIC_URL + "/run"
print()
print("Endpoint URL (paste into the Tool tab -> Configure endpoint):")
print("   ", ENDPOINT_URL)
```

```python
import ast, json, os, urllib.error, urllib.request

# Built from the tool_interface declared for @localtester/energy-local. This cannot hand-verify a real
# answer the way the Hub's own reference wrapper does — it checks that the contract
# is honoured, not that any particular answer is correct.
tool_names = ['calculate_levelized_cost_of_energy']

for name in tool_names:
    envelope = {"powerai": "1", "tool": name, "input": {}}
    request = urllib.request.Request(
        ENDPOINT_URL, data=json.dumps(envelope).encode(),
        headers={"Content-Type": "application/json"},
    )
    try:
        reply = json.load(urllib.request.urlopen(request, timeout=90))
    except urllib.error.HTTPError as err:
        reply = json.load(err)
    assert "ok" in reply, f"{name}: response has no 'ok' field — {reply}"
    if reply["ok"]:
        print(f"{name} -> ok (empty input was accepted)")
    else:
        error = reply.get("error", {})
        assert error.get("code") != "unknown_tool" and "unknown tool" not in str(
            error.get("message", "")
        ).lower(), (
            f"{name} was not recognised — check the spelling against its "
            "tool_interface.name"
        )
        print(f"{name} -> recognised, rejected empty input ({error.get('message')})")

print()
print("endpoint speaks the contract for every declared tool.")

# The shape check above sends an empty input, which a wrapper that works and one that
# crashes on every valid input both reject. A file can therefore pass it and still be
# broken, and that was most of what went wrong in drafted ones. So if the endpoint file
# carries SAMPLE_INPUTS (one realistic input per tool), send each one and print what the
# primitive really computes. Read with literal_eval, never imported: it is data.
samples = {}
try:
    source = open(os.path.join(WORKDIR, MODULE + ".py")).read()
    for statement in ast.parse(source).body:
        if isinstance(statement, ast.Assign) and any(
            getattr(target, "id", "") == "SAMPLE_INPUTS" for target in statement.targets
        ):
            samples = ast.literal_eval(statement.value)
except Exception as exc:
    print(f"could not read SAMPLE_INPUTS ({exc}); skipping the sample run")

failed = []
for name, sample in samples.items():
    envelope = {"powerai": "1", "tool": name, "input": sample}
    request = urllib.request.Request(
        ENDPOINT_URL, data=json.dumps(envelope).encode(),
        headers={"Content-Type": "application/json"},
    )
    try:
        reply = json.load(urllib.request.urlopen(request, timeout=90))
    except urllib.error.HTTPError as err:
        reply = json.load(err)
    shown = json.dumps(reply)[:600]
    if reply.get("ok"):
        print(f"{name} <- {sample}")
        print(f"   computed: {shown}")
    else:
        failed.append(name)
        print(f"{name} <- {sample}")
        print(f"   FAILED: {shown}")

assert not failed, (
    f"the sample input was rejected for {failed} — a valid input should compute, so "
    "the endpoint file is probably wrong. Fix it (or SAMPLE_INPUTS) and run again."
)

# An agent's tool layer sends every number as a float: 25 arrives as 25.0. A file that
# accepts the sample as written can still reject every real call, so send the same sample
# again with whole numbers as floats.
def as_floats(value):
    if isinstance(value, bool):
        return value
    if isinstance(value, int):
        return float(value)
    if isinstance(value, dict):
        return {key: as_floats(item) for key, item in value.items()}
    if isinstance(value, list):
        return [as_floats(item) for item in value]
    return value

float_failed = []
for name, sample in samples.items():
    envelope = {"powerai": "1", "tool": name, "input": as_floats(sample)}
    request = urllib.request.Request(
        ENDPOINT_URL, data=json.dumps(envelope).encode(),
        headers={"Content-Type": "application/json"},
    )
    try:
        reply = json.load(urllib.request.urlopen(request, timeout=90))
    except urllib.error.HTTPError as err:
        reply = json.load(err)
    if not reply.get("ok"):
        float_failed.append(name)
        print(f"{name} <- {as_floats(sample)}")
        print(f"   FAILED with whole numbers as floats: {json.dumps(reply)[:600]}")

assert not float_failed, (
    f"{float_failed} computed the sample as written but rejected the same numbers as "
    "floats (25 sent as 25.0). Agents send numbers that way, so every real call would "
    "fail: accept a float with no fractional part for an integer input."
)
print()
if samples:
    print("Every sample input computed, also with whole numbers as floats. Check that the")
    print("numbers above look right: only you can say whether they are correct, and this")
    print("check cannot.")
else:
    print("No SAMPLE_INPUTS in the endpoint file, so only the contract's shape was checked.")
    print("Whether its answers are correct is on you, not this check.")
```
